# Distribuição amostral e intervalo de confiança

```python
# Visualizar todos os salários
salarios = dados['SALARIO']
print(salarios )
```

```python
# Média de amostra de salários
media_amostral = np.mean(salarios)
media_amostral
```

```python
# Calculando desvio padrão da amostra
desvio_amostral = np.std(salarios)
desvio_amostral
```

```python
# Definição de nível de confiança
nivel_confianca = 0.95
```

```python
# Verificando o tamanho da amostra
tamanho_amostra  = len(salarios)
tamanho_amostra
```

```python
# importando 1 módulo da biblioteca scipy
from scipy import stats
```

```python
# Usando funçao sem, para calcular o erro médio padrão da amostra
erro_padrao = stats.sem(salarios)
erro_padrao
```

```python
# Calculando intervalo de confiança
intervalo_confianca = stats.t.interval(nivel_confianca, tamanho_amostra-1, loc=media_amostral, scale=erro_padrao)
intervalo_confianca
```

```python
# Importando função chi2
from scipy.stats import chi2_contingency
```

```python
def cramer_coeficiente(coluna1, coluna2):
  tabela_cruzada = np.array(pd.crosstab(coluna1, coluna2))
  chi2 = chi2_contingency(tabela_cruzada)[0]
  soma = np.sum(tabela_cruzada)
  mini = min(tabela_cruzada.shape)-1
  cramer = np.sqrt(chi2/(soma*mini))
  return cramer
```

```python
cramer_coeficiente(coluna1=dados['COR/RACA/ETNIA'], coluna2=dados['NIVEL DE ENSINO'])
```

```python
# Verificando quais estados brasileiros estão cadastrados, sem que apareça estados que não tem dados cadastrados
list_estados = list(dados['UF ONDE MORA'].unique())
```

```python
query = '''SELECT Municipios_Brasileiros.Estado, AVG(Municipio_Status.renda) FROM  Municipios_Brasileiros
        INNER JOIN Municipio_Status ON Municipios_Brasileiros.municipio_ID = Municipio_Status.municipio_ID
        WHERE Municipios_Brasileiros.Estado IN ({})
        GROUP BY Municipios_Brasileiros.Estado'''.format(','.join(['?' for _ in list_estados]))
```

```python
print(query)
```

```python
# Pedindo para o pandas ler o SQL e trazer a lista de estados e média de renda
estados_renda = pd.read_sql(query, con=conexao, params=list_estados)
```

```python
# Renomeando colunas para evitar conflitos e padronizar para o merge
print("Colunas antes da renomeação:\n", dados.columns)

# Obtém a lista atual de colunas
cols = dados.columns.tolist()

# As colunas em questão são 'ESTADO ONDE MORA' e 'UF ONDE MORA' no DataFrame original.
# Como o DataFrame atual tem duas colunas simplesmente chamadas 'ESTADO',
# vamos renomeá-las por índice, assumindo a ordem original.
# A coluna em index 11 era 'ESTADO ONDE MORA' (cidade/nome completo do estado)
# A coluna em index 12 era 'UF ONDE MORA' (sigla do estado, que será usada para o merge)

# Verifica se as colunas nos índices esperados são de fato 'ESTADO' antes de renomear
if cols[11] == 'ESTADO' and cols[12] == 'ESTADO':
    cols[11] = 'CIDADE ONDE MORA'
    cols[12] = 'Estado'
    dados.columns = cols
elif 'UF ONDE MORA' in dados.columns and 'ESTADO ONDE MORA' in dados.columns:
    # Fallback caso os nomes originais ainda existam (menos provável, mas para robustez)
    dados.rename(columns={'UF ONDE MORA':'Estado', 'ESTADO ONDE MORA':'CIDADE ONDE MORA'}, inplace=True)

print("Colunas após a renomeação:\n", dados.columns)
```

```python
dados.columns
```

```python
# Fazendo merge de estados renda através da coluna de estado.
# O how(como) é a coluna da esquerda (left)
dados.merge(estados_renda, on="Estado", how='left')
```

```python
# Atribuindo o resultado do merge a um novo DataFrame e exibindo as primeiras linhas
dados_merge = dados.merge(estados_renda, on="Estado", how='left')
display(dados_merge.head())
```
