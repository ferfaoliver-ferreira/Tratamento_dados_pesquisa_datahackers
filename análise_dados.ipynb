# Distribuição amostral e intervalo de confiança

```python
# Visualizar todos os salários
salarios = dados['SALARIO']
print(salarios )
```

```python
# Média de amostra de salários
media_amostral = np.mean(salarios)
media_amostral
```

```python
# Calculando desvio padrão da amostra
desvio_amostral = np.std(salarios)
desvio_amostral
```

```python
# Definição de nível de confiança
nivel_confianca = 0.95
```

```python
# Verificando o tamanho da amostra
tamanho_amostra  = len(salarios)
tamanho_amostra
```

```python
# importando 1 módulo da biblioteca scipy
from scipy import stats
```

```python
# Usando funçao sem, para calcular o erro médio padrão da amostra
erro_padrao = stats.sem(salarios)
erro_padrao
```

```python
# Calculando intervalo de confiança
intervalo_confianca = stats.t.interval(nivel_confianca, tamanho_amostra-1, loc=media_amostral, scale=erro_padrao)
intervalo_confianca
```

# Featuring engineering


```python
# Função de preencher nível
def preencher_nivel(gestor, nivel):
  if gestor ==1:
    return 'Pessoa Gestora'
  else:
    return nivel
```

```python
# Aplicando função apply do pandas para aplicar função específica para linhas ou colunas da tabela
dados['NOVO_NIVEL'] = dados.apply(lambda x: preencher_nivel(x['GESTOR?'], x['NIVEL']), axis=1)
dados['NOVO_NIVEL'].value_counts()
```

```python
# Importando função chi2
from scipy.stats import chi2_contingency
```

```python
def cramer_coeficiente(coluna1, coluna2):
  tabela_cruzada = np.array(pd.crosstab(coluna1, coluna2))
  chi2 = chi2_contingency(tabela_cruzada)[0]
  soma = np.sum(tabela_cruzada)
  mini = min(tabela_cruzada.shape)-1
  cramer = np.sqrt(chi2/(soma*mini))
  return cramer
```

```python
cramer_coeficiente(coluna1=dados['COR/RACA/ETNIA'], coluna2=dados['NIVEL DE ENSINO'])
```
